# A/B Testing EduPath AI

## Latar Belakang

EduPath AI merupakan platform pembelajaran adaptif yang menyediakan rekomendasi materi berdasarkan aktivitas belajar siswa.

Untuk mengevaluasi efektivitas fitur rekomendasi tersebut, dilakukan simulasi A/B Testing.

## Tujuan

Mengetahui apakah penggunaan sistem rekomendasi pembelajaran berpotensi meningkatkan performa siswa dibandingkan tanpa sistem rekomendasi.

## Hipotesis

### H0 (Hipotesis Nol)

Tidak terdapat perbedaan performa siswa antara kelompok yang menggunakan rekomendasi AI dan kelompok yang tidak menggunakan rekomendasi AI.

### H1 (Hipotesis Alternatif)

Terdapat perbedaan performa siswa antara kelompok yang menggunakan rekomendasi AI dan kelompok yang tidak menggunakan rekomendasi AI.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind
```

```python
np.random.seed(42)

# Kelompok tanpa rekomendasi AI
group_A = np.random.normal(
    loc=70,
    scale=10,
    size=100
)

# Kelompok menggunakan rekomendasi AI
group_B = np.random.normal(
    loc=78,
    scale=10,
    size=100
)
```

```python
summary = pd.DataFrame({
    "Kelompok": ["A (Tanpa AI)", "B (Dengan AI)"],
    "Jumlah Siswa": [len(group_A), len(group_B)],
    "Rata-rata Nilai": [round(group_A.mean(),2), round(group_B.mean(),2)],
    "Standar Deviasi": [round(group_A.std(),2), round(group_B.std(),2)]
})

summary
```

```python
plt.figure(figsize=(8,5))

plt.hist(group_A, alpha=0.6, label="Tanpa Rekomendasi AI")
plt.hist(group_B, alpha=0.6, label="Dengan Rekomendasi AI")

plt.title("Distribusi Nilai Siswa")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.legend()

plt.show()
```

```python
t_stat, p_value = ttest_ind(
    group_A,
    group_B
)

print("T-Statistic :", round(t_stat,4))
print("P-Value     :", p_value)
```

```python
alpha = 0.05

print("Alpha =", alpha)

if p_value < alpha:
    print("\nKeputusan:")
    print("Tolak H0")

    print("\nInterpretasi:")
    print(
        "Terdapat perbedaan signifikan antara kelompok "
        "yang menggunakan rekomendasi AI dan kelompok "
        "yang tidak menggunakan rekomendasi AI."
    )

else:
    print("\nKeputusan:")
    print("Gagal Menolak H0")

    print("\nInterpretasi:")
    print(
        "Tidak terdapat perbedaan signifikan antara kedua kelompok."
    )
```

# Hasil Pengujian

Berdasarkan hasil Independent Sample T-Test, diperoleh nilai p-value yang lebih kecil dari tingkat signifikansi (α = 0,05).

Karena p-value < 0,05, maka hipotesis nol (H0) ditolak.

Hal ini menunjukkan bahwa terdapat perbedaan performa yang signifikan antara kelompok yang menggunakan rekomendasi AI dan kelompok yang tidak menggunakan rekomendasi AI.

# Kesimpulan

Berdasarkan simulasi A/B Testing yang dilakukan, kelompok yang menggunakan rekomendasi AI menunjukkan performa yang lebih baik dibandingkan kelompok tanpa rekomendasi AI.

Hasil pengujian statistik menunjukkan bahwa perbedaan tersebut signifikan secara statistik, sehingga fitur rekomendasi pembelajaran berpotensi memberikan dampak positif terhadap hasil belajar siswa.
