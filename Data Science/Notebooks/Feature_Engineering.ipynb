```python
import pandas as pd
import numpy as np

df = pd.read_csv("cleaned_student_data.csv")
```

```python
df["engagement_score"] = (
    df["raisedhands"]
    + df["VisITedResources"]
    + df["AnnouncementsView"]
    + df["Discussion"]
)
```

```python
df["activity_level"] = pd.cut(
    df["engagement_score"],
    bins=[0,100,200,400],
    labels=["Low","Medium","High"]
)
```

```python
df["absence_risk"] = np.where(
    df["StudentAbsenceDays"]=="Above-7",
    1,
    0
)
```

```python
df.to_csv(
    "student_feature_engineered.csv",
    index=False
)
```
